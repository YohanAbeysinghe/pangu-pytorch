### Surface Variables

```python
import pandas as pd

# Define the file paths
file_path_acc = "/pfs/lustrep1/scratch/project_462000472/akhtar/climate_modeling/pangu-data/non_cropped_model/results/test_original_pangu/test/csv/acc_surface.csv"
file_path_rmse = "/pfs/lustrep1/scratch/project_462000472/akhtar/climate_modeling/pangu-data/non_cropped_model/results/test_original_pangu/test/csv/rmse_surface.csv"

# Load the CSV files
df_acc = pd.read_csv(file_path_acc)
df_rmse = pd.read_csv(file_path_rmse)

# Drop the first column (dates)
df_acc = df_acc.drop(columns=df_acc.columns[0])
df_rmse = df_rmse.drop(columns=df_rmse.columns[0])

# Calculate the means
averages_acc = df_acc.mean()
averages_rmse = df_rmse.mean()

# Combine into a table
df_results = pd.DataFrame({'Average ACC': averages_acc, 'Average RMSE': averages_rmse})

# Print as a table
print(df_results)
```

### Upper Variables

```python
import pandas as pd
import os

# Define the folder path
folder_path = "/pfs/lustrep1/scratch/project_462000472/akhtar/climate_modeling/pangu-data/non_cropped_model/results/test_original_pangu/test/csv/"

# List of variable names
variables = ['z', 'q', 't', 'u', 'v']

# Initialize dictionaries to hold the data for ACC and RMSE
acc_data = {}
rmse_data = {}

# Load CSV files and process
for var in variables:
    acc_data[var] = pd.read_csv(os.path.join(folder_path, f"acc_upper_{var}.csv")).iloc[:, 1:]  # Drop the first column
    rmse_data[var] = pd.read_csv(os.path.join(folder_path, f"rmse_upper_{var}.csv")).iloc[:, 1:]  # Drop the first column

# Calculate the means for each variable's ACC and RMSE
averages_acc = {var: acc_data[var].mean() for var in variables}
averages_rmse = {var: rmse_data[var].mean() for var in variables}

# Create DataFrame for ACC values
df_acc = pd.DataFrame({
    f'{var.upper()} - ACC': averages_acc[var] for var in variables
})

# Create DataFrame for RMSE values
df_rmse = pd.DataFrame({
    f'{var.upper()} - RMSE': averages_rmse[var] for var in variables
})

# Add the overall average row for ACC
df_acc.loc['Average'] = df_acc.mean()

# Add the overall average row for RMSE
df_rmse.loc['Average'] = df_rmse.mean()

# Add multi-index headers for ACC and RMSE DataFrames
df_acc.columns = pd.MultiIndex.from_tuples([
    (var.upper(), 'ACC') for var in variables
])

df_rmse.columns = pd.MultiIndex.from_tuples([
    (var.upper(), 'RMSE') for var in variables
])

# Print the results as two separate tables
print("ACC Table:")
print(df_acc)

print("\nRMSE Table:")
print(df_rmse)
```

```python
import pandas as pd
import os
from tabulate import tabulate  # Import tabulate

# Define the folder path
folder_path = "/pfs/lustrep1/scratch/project_462000472/akhtar/climate_modeling/pangu-data/non_cropped_model/results/test_original_pangu/test/csv/"

# List of variable names
variables = ['z', 'q', 't', 'u', 'v']

# Initialize dictionaries to hold the data for ACC and RMSE
acc_data = {}
rmse_data = {}

# Load CSV files and process
for var in variables:
    acc_data[var] = pd.read_csv(os.path.join(folder_path, f"acc_upper_{var}.csv")).iloc[:, 1:]  # Drop the first column
    rmse_data[var] = pd.read_csv(os.path.join(folder_path, f"rmse_upper_{var}.csv")).iloc[:, 1:]  # Drop the first column

# Calculate the means for each variable's ACC and RMSE
averages_acc = {var: acc_data[var].mean() for var in variables}
averages_rmse = {var: rmse_data[var].mean() for var in variables}

# Create DataFrame for ACC values
df_acc = pd.DataFrame({
    f'{var.upper()} - ACC': averages_acc[var] for var in variables
})

# Create DataFrame for RMSE values
df_rmse = pd.DataFrame({
    f'{var.upper()} - RMSE': averages_rmse[var] for var in variables
})

# Add the overall average row for ACC
df_acc.loc['Average'] = df_acc.mean()

# Add the overall average row for RMSE
df_rmse.loc['Average'] = df_rmse.mean()

# Directly set column headers to variable names
df_acc.columns = [var.upper() for var in variables]
df_rmse.columns = [var.upper() for var in variables]

# Print the results as two separate tables with compact formatting
print("ACC Table:")
print(tabulate(df_acc, headers='keys', tablefmt='fancy_grid', showindex=True, numalign="center", stralign="center"))

print("\nRMSE Table:")
print(tabulate(df_rmse, headers='keys', tablefmt='fancy_grid', showindex=True, numalign="center", stralign="center"))
```
