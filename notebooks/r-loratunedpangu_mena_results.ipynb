```python
import pandas as pd
import os
from tabulate import tabulate  # Import tabulate

# Define the folder path
folder_path = "/pfs/lustrep1/scratch/project_462000472/akhtar/climate_modeling/pangu-data/non_cropped_with_pm2.5/results/Feb_22_2/csv"

# List of variable names
variables = ['z', 'q', 't', 'u', 'v']

# Initialize dictionaries to hold the data for ACC and RMSE
acc_data = {}
rmse_data = {}

# Load CSV files and process
for var in variables:
    acc_data[var] = pd.read_csv(os.path.join(folder_path, f"acc_upper_{var}.csv")).iloc[:, 1:]  # Drop the first column
    rmse_data[var] = pd.read_csv(os.path.join(folder_path, f"rmse_upper_{var}.csv")).iloc[:, 1:]  # Drop the first column

# Calculate the means for each variable's ACC and RMSE
averages_acc = {var: acc_data[var].mean() for var in variables}
averages_rmse = {var: rmse_data[var].mean() for var in variables}

# Create DataFrame for ACC values
df_acc = pd.DataFrame({
    f'{var.upper()} - ACC': averages_acc[var] for var in variables
})

# Create DataFrame for RMSE values
df_rmse = pd.DataFrame({
    f'{var.upper()} - RMSE': averages_rmse[var] for var in variables
})

# Add the overall average row for ACC
df_acc.loc['Average'] = df_acc.mean()

# Add the overall average row for RMSE
df_rmse.loc['Average'] = df_rmse.mean()

# Directly set column headers to variable names
df_acc.columns = [var.upper() for var in variables]
df_rmse.columns = [var.upper() for var in variables]

# Print the results as two separate tables with compact formatting
print("ACC Table:")
print(tabulate(df_acc, headers='keys', tablefmt='fancy_grid', showindex=True, numalign="center", stralign="center"))

print("\nRMSE Table:")
print(tabulate(df_rmse, headers='keys', tablefmt='fancy_grid', showindex=True, numalign="center", stralign="center"))
```
