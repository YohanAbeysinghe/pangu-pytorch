```python
import pandas as pd

# Define the file paths
file_path_acc = "/pfs/lustrep1/scratch/project_462000472/akhtar/climate_modeling/pangu-data/non_cropped_with_pm2.5/results/train_with_pm_5years_3/csv/acc_surface.csv"
file_path_rmse = "/pfs/lustrep1/scratch/project_462000472/akhtar/climate_modeling/pangu-data/non_cropped_with_pm2.5/results/train_with_pm_5years_3/csv/rmse_surface.csv"

# Load the CSV files
df_acc = pd.read_csv(file_path_acc)
df_rmse = pd.read_csv(file_path_rmse)

# Drop the first column (dates)
df_acc = df_acc.drop(columns=df_acc.columns[0])
df_rmse = df_rmse.drop(columns=df_rmse.columns[0])

# Calculate the means
averages_acc = df_acc.mean()
averages_rmse = df_rmse.mean()

# Combine into a table
df_results = pd.DataFrame({'Average ACC': averages_acc, 'Average RMSE': averages_rmse})

# Print as a table
print(df_results)
```
