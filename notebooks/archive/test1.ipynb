```python
import os
cams_file_path = "/l/users/fahad.khan/akhtar/Pangu/data/data_prep/pm/2019/cams_201901_025_renamed.nc"
surface_file_path = "/l/users/fahad.khan/akhtar/Pangu/data/data_prep/merged_proper/surface_201901.nc"

dataset1 = xr.open_dataset(cams_file_path)  # Dataset with pm2p5
dataset2 = xr.open_dataset(surface_file_path)  # Dataset with u10, v10, t2m, msl

# Merge the datasets
merged_dataset = xr.merge([dataset2, dataset1])

# Define the output file path
output_file_path = os.path.join("/l/users/fahad.khan/akhtar/Pangu/pangu-pytorch/notebooks/output.nc")

# Save the merged dataset
merged_dataset.to_netcdf(output_file_path)
```
